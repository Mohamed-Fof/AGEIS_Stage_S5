## Ajout de la variable HDI


```python
data3 = pd.read_excel('hdr-data.xlsx')
data3.head()
```

```python
data3 = data3.drop(columns=['dimension','yearStr','value','note'])
data3.head(10)
```

```python
data3['indicator'].unique()
```

```python
data3['index'].unique()
```

```python
# 
data3 = data3[data3["index"] == "Human Development Index"]

#
df3 = data3.pivot_table(
    index=["countryIsoCode", "country", "year"],
    values="actualValue"
).reset_index()


df3 = df3.rename(columns={
    "countryIsoCode": "Code_Pays",
    "country": "Pays",
    "year": "Années",
    "Human Development Index (value)": "IDH",
    'actualValue': "IDH"
})


dico_pays = {
    "South Africa": "Afrique du Sud",
    "Germany": "Allemagne",
    "Brazil": "Brésil",
    "Canada": "Canada",
    "Chile": "Chili",
    "Hong Kong, China (SAR)": "Chine",
    "Korea (Republic of)": "Corée du Sud",
    "Côte d'Ivoire": "Côte d'Ivoire",
    "Spain": "Espagne",
    "France": "France",
    "India": "Inde",
    "Indonesia": "Indonésie",
    "Iran (Islamic Republic of)": "Iran",
    "Japan": "Japon",
    "Morocco": "Maroc",
    "Mexico": "Mexique",
    "Nigeria": "Nigeria",
    "Peru": "Pérou",
    "United Kingdom": "Royaume-Uni",
    "Russian Federation": "Russie",
    "Switzerland": "Suisse",
    "Sweden": "Suède",
    "Senegal": "Sénégal",
    "Togo": "Togo",
    "Türkiye": "Turquie",
    "United States": "États-Unis"
}

df3["Pays"] = df3["Pays"].replace(dico_pays)
df3.head(50)
```

```python
df3.isnull().sum()
```

```python
df3.shape
```

```python
df3.columns
```

```python
df3['Pays'].unique()
```

```python
df3 = df3.sort_values(by=["Pays", "Années"])
df3 = df3.reset_index(drop=True)
df3["Pays"].unique()
```

```python
df3.to_csv("DataIDH.csv",index=False, sep=';')
```

```python
pays_developed = [
    "Allemagne", "Canada", "Suisse", "France", "Royaume-Uni",
    "États-Unis", "Suède", "Japon", "Corée du Sud", "Espagne",
    "Russie", "Chine", "Espagne"
]

df3["Statut"] = df3["Pays"].apply(
    lambda x: "Developed" if x in pays_developed else "Developing"
)

df3.head(50)
```

```python
df3.isna().sum()
```

```python
df3= df3[~df3["Années"].isin([2022, 2023])]
df3['Années'].unique()
```

```python
# Sélectionner seulement les colonnes utiles
cols_to_keep = [
    'Années',
    'Pays',
    'Code_Pays',
    "Chômage, femmes (% de la population active féminine)",
    "Chômage, hommes (% de la population active masculine)",
    "Mortalité prématurée par maladies cardiovasculaires, cancer, diabète ou maladies respiratoires chroniques (femmes, 30-70 ans)",
    "Mortalité prématurée par maladies cardiovasculaires, cancer, diabète ou maladies respiratoires chroniques (hommes, 30-70 ans)",
    "Émissions de CO2 par habitant (tonnes)"
]

df = df4[cols_to_keep]

df = df.rename(columns={
    "Chômage, femmes (% de la population active féminine)": "Chomage_femmes",
    "Chômage, hommes (% de la population active masculine)": "Chomage_hommes",
    "Mortalité prématurée par maladies cardiovasculaires, cancer, diabète ou maladies respiratoires chroniques (femmes, 30-70 ans)": "Mort_prem_F",
    "Mortalité prématurée par maladies cardiovasculaires, cancer, diabète ou maladies respiratoires chroniques (hommes, 30-70 ans)": "Mort_prem_H",
    "Émissions de CO2 par habitant (tonnes)": "CO2_habitant"
})
df.head()
```

```python
idx = df_final.columns.get_loc("VIH_SIDA") + 1
cols_to_move = df_final.columns[-5:].tolist()
for i, col in enumerate(cols_to_move):
    # Supprimer la colonne et récupérer les valeurs
    vals = df_final.pop(col)
    # Insérer après HID (idx+i pour avancer à chaque insertion)
    df_final.insert(loc=idx + i, column=col, value=vals)
df_final.columns
```
