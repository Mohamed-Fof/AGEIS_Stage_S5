```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt 
```

```python
data = pd.read_csv("~/Desktop/Stage/Datacancer.csv")
data.head()
```

```python
# 1️⃣ Sélectionner les colonnes qui t'intéressent
colonnes_utiles = ["year", "location_name", "cause_name", "val"]

data1 = data[colonnes_utiles].copy()

# 2️⃣ Renommer les colonnes
data1 = data1.rename(columns={
    "year": "Années",
    "location_name": "Pays",
    "cause_name": "Cause",
    "val": "Taux"
})
data1.head()
```

```python
data1.shape
```

# 

```python
# 1) Optionnel : s'assurer que les années sont bien des entiers
data1["Années"] = data1["Années"].astype(int)

# 2) Pivot : chaque cause devient une colonne
data2 = data1.pivot_table(
    index=["Années", "Pays"],   # lignes = combinaison année + pays
    columns="Cause",            # colonnes = les différentes causes
    values="Taux",              # valeurs dans les cellules
    aggfunc="first"             # ou "sum", "mean"... si doublons possibles
)

# 3) Remettre l'index à plat
data2 = data2.reset_index()

# 4) (Optionnel) enlever le nom de l'index des colonnes
data2.columns.name = None

# 5) (Optionnel) remplacer les NaN par 0 ou "X" si tu préfères
# df_wide = df_wide.fillna("X")

data2.head()
```

```python
data2 = data2.rename(columns={ "Cancer de la thyroïde": "Cancer_thyroïde", 
                        "Cancer de l’estomac": "Cancer_Estomac", 
                        "Cancer de l’utérus": "Cancer_Utérus",
                        "Cancer du foie": "Cancer_Foie",
                        "Cancer du sein": "Cancer_Sein",
                        "Cancer du testicule":"Cancer_Testicule",
                       })
data2.head()
```

```python
data2 = data2.sort_values(["Pays", "Années"]).reset_index(drop=True)
data2.head()
```
